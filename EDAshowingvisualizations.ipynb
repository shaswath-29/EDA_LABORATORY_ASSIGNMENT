The Iris dataset contains measurements for different iris species; perform exploratory data analysis and present the results using visualizations.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)
```

```python
df['species'] = [iris.target_names[i] for i in iris.target]
```

```python
print("First 5 rows:")
df.head()
```

```python
print("\nDataset Information:")
print(df.info())
```

```python
print("\nShape of dataset:")
df.shape
```

```python
print("\nMissing values:")
df.isnull().sum()
```

```python
print("\nSummary Statistics:")
df.describe()
```

```python
print("\nSpecies Count:")
df['species'].value_counts()
```

```python
df.hist(figsize=(10, 8), bins=15)
plt.suptitle("Distribution of Iris Features")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='species',
    y='sepal length (cm)'
)
plt.title("Sepal Length by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x='petal length (cm)',
    y='petal width (cm)',
    hue='species',
    s=80
)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x='sepal length (cm)',
    y='sepal width (cm)',
    hue='species',
    s=80
)

plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.show()
```

```python
sns.pairplot(
    df,
    hue='species'
)
plt.suptitle("Pairwise Relationships in Iris Dataset",y=1.02)
plt.show()
```

```python
correlation = df.iloc[:, 0:4].corr()
print("\nCorrelation Matrix:")
correlation
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
plt.title("Correlation Heatmap")
plt.show()
```
