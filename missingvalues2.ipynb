```python

import pandas as pd;
import numpy as np;
data = np.arange(15, 30).reshape(5, 3)
dfx = pd.DataFrame(data, index=['chaitanya', 'hasith', 'praneeth',
'alex', 'adam'], columns=['CN', 'DBMS', 'EDA'])
dfx
```

```python
dfx['PROB'] = np.nan
dfx.loc['shaswath'] = np.arange(15, 19)
dfx.loc['wasim'] = np.nan
dfx['PROB'] = np.nan
dfx.loc['chaitanya', 'HSM'] = 20.
dfx
```

```python
print(dfx.isnull())
print("notnull()")
print(dfx.notnull())
print("NaNs per column")
print(dfx.isnull().sum())
print("Total missing values in DataFrame")
print(dfx.isnull().sum().sum())
print("Count of reported values")
print(dfx.count())
```

```python
print("Valid values in HSM")
print(dfx.HSM[dfx.HSM.notnull()])
print("HSM series after dropna()")
print(dfx.HSM.dropna())
print("dfx.dropna() (drops rows with any NaN)")
print(dfx.dropna())
print("dfx.dropna(how='all')")
print(dfx.dropna(how='all'))
print("dfx.dropna(how='all', axis=1)")
print(dfx.dropna(how='all', axis=1))
print("dfx.dropna(thresh=10, axis=1)")
print(dfx.dropna(thresh=10, axis=1))
```

```python
ar1 = np.array([85, 90, np.nan, 95])
ser1 = pd.Series(ar1)
print("NumPy Mean (returns NaN):", ar1.mean())
print("Pandas Mean (ignores NaN):", ser1.mean())
ser2 = dfx.HSM
print("Total HSM score (sum)")
print(ser2.sum())
print("Average HSM score (mean)")
print(ser2.mean())
print("Cumulative sum of HSM scores")
print(ser2.cumsum())
```

```python
filledDf = dfx.fillna(0)
print("DataFrame filled with 0")
print(filledDf)
print("Original Means")
print(dfx.mean())
print("Means After Filling with 0")
print(filledDf.mean())
print("Forward Fill (ffill) on HSM")
print(dfx.HSM.ffill())
print("Backward Fill (bfill) on HSM")
print(dfx.HSM.bfill())
```
