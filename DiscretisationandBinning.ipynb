### Loading a different dataset (e.g., a generic CSV)

```python
import pandas as pd
import numpy as np
marks = [35,42,55,67,72,81,49,90,76,64,58,39,95,88,73,61,47,52,79,84]
print("Original Marks:")
print(marks)
```

```python
bins = [0,40,60,80,100]
category = pd.cut(marks,bins)
print("\nMarks after binning:")
print(category)
```

```python
print("\nNumber of students in each category:")
print(pd.Series(category).value_counts())
```

```python
labels = ["Fail","Average","Good","Excellent"]
category_labels = pd.cut(marks,bins,labels = labels)
print("\nMarks with labels:")
print(category_labels)
```

```python
random_marks = np.random.randint(0,101,40)
print("\nRandom Marks:")
print(random_marks)
equal_bins = pd.cut(random_marks,5,precision=2)
print("\nEqual Width Bins:")
print(equal_bins)
print("\nCount in Equal Width Bins:")
print(pd.Series(equal_bins).value_counts())
```

```python
random_scores = np.random.randint(0,101,200)
quartiles = pd.qcut(random_scores,4)
print("\nQuantile Bins:")
print(quartiles)
print("\nCount in Quantile Bins:")
print(pd.Series(quartiles).value_counts())
```
